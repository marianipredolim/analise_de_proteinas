**Primeiro bloco:** cria uma função genérica pra identificar o arquivo, ler o título e a sequência em si.

**Segundo bloco:** inicia o loop "for" com o condicional "if" que vai ser responsável por ler linha a linha pra entender se é parte da sequência ou se é uma nova por meio dos comandos abaixo:
- **Strip:** trata as quebras de linhas
- **Startswith:** identifica o inicio de sequência e retira o ">"
- **yield:** devolve na saída o que já foi analisado e finalizado
  essa parte: "header = linha[1:]" atualiza o nome da sequência quando a última acabou, senão adiciona as outras linhas.

Em resumo, identifica onde tem um novo nome de sequência e tudo que não é nome ele trata e junta considerando ser a sequência em si.


**Terceiro bloco:** "chama" a função que criamos e aplica para o nosso .fasta, para isso funcionar só precisamos colocar o nome do arquivo onde estava a váriavel genérica "arquivo" -> ler_fasta(arquivo)

E dicionei os "print" só para nos mostrar as informações que nossa função leu.
ps: coloquei pra mostrar parcial as sequências pra não ficar muito poluído



```python
### FUNÇÃO PARA LER .FASTA

#Primeiro bloco
def ler_fasta(arquivo):
    with open(arquivo, 'r') as f:
        header = ""
        sequencia = ""

#Segundo bloco
        for linha in f:
            linha = linha.strip()
            if linha.startswith('>'):
                if header:
                    yield (header, sequencia)
                header = linha[1:]
                sequencia = ""
            else:
                sequencia += linha

        if header:
            yield (header, sequencia)

#Terceiro Bloco
for header, sequencia in ler_fasta("uniprotkb_ctenitoxin_2025_11_13.fasta"):
      print(f">>{header}")
      print(f"Tamanho: {len(sequencia)}")
      print(f"Sequência: {sequencia[:30]}...")
      print()
```
